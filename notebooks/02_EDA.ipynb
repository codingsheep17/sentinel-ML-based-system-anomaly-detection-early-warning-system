Doing EDA on prceessed data

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("D:/Desktop/sentinelml/data/processed/processed.csv", parse_dates=["timestamp"])
```

```python
sns.set_theme(style="darkgrid")
#figure size
plt.figure(figsize=(15, 5))

#plot the time series using Seaborn
#we simply pass the dataframe and tell it which columns to use for X and Y
sns.lineplot(data=df, x="timestamp", y="cpu_utilization", color="crimson", linewidth=1.2)

#add titles and axis labels
plt.title("AWS EC2 CPU Utilization Over Time", fontsize=14, fontweight="bold")
plt.xlabel("Time", fontsize=12)
plt.ylabel("CPU Utilization", fontsize=12)

plt.tight_layout()
plt.show()
```
